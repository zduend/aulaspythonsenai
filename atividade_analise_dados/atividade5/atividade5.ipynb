```python
import pandas as pd
import numpy as np

# Carregamento dos dados
df = pd.read_csv("vendas_mercados.csv", sep="\t")

# Verificando a base
print(df.head())
print(df.dtypes)
```

```python
# QUESTÃO 1
df["Lucro"] = (
    df["Preco_Unitario"] - df["Custo_Unitario"]
) * df["Quantidade"]
lucro_por_mercado = df.groupby("Mercado")["Lucro"].sum()
print("\n--- LUCRO POR MERCADO ---")
print(lucro_por_mercado)
print("\n--- MERCADO COM MAIOR LUCRO ---")
print(lucro_por_mercado.idxmax())
print(f"R$ {lucro_por_mercado.max():.2f}")
```

```python
# QUESTÃO 2 - Forma de pagamento mais utilizada por região

pagamentos = df.groupby(
    ["Regiao", "Forma_Pagamento"]
).size()
print("\n--- FORMAS DE PAGAMENTO POR REGIÃO ---")
print(pagamentos)

# Identificando os pagamentos mais utilizados, incluindo empates
maior_frequencia = pagamentos.groupby(level=0).transform("max")

pagamentos_mais_usados = pagamentos[
    pagamentos == maior_frequencia
]

print("\n--- PAGAMENTOS MAIS UTILIZADOS POR REGIÃO ---")
print(pagamentos_mais_usados)
```
