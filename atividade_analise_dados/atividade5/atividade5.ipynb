```python
import pandas as pd
import numpy as np

# Carregamento dos dados
df = pd.read_csv("vendas_mercados.csv", sep="\t")

# Verificando a base
print(df.head())
print(df.dtypes)
```

```python
# QUESTÃO 1
df["Lucro"] = (
    df["Preco_Unitario"] - df["Custo_Unitario"]
) * df["Quantidade"]
lucro_por_mercado = df.groupby("Mercado")["Lucro"].sum()
print("\n--- LUCRO POR MERCADO ---")
print(lucro_por_mercado)
print("\n--- MERCADO COM MAIOR LUCRO ---")
print(lucro_por_mercado.idxmax())
print(f"R$ {lucro_por_mercado.max():.2f}")
```

```python
# QUESTÃO 2

pagamentos = df.groupby(
    ["Regiao", "Forma_Pagamento"]
).size()
print("\n--- FORMAS DE PAGAMENTO POR REGIÃO ---")
print(pagamentos)

# Identificando os pagamentos mais utilizados, incluindo empates
maior_frequencia = pagamentos.groupby(level=0).transform("max")

pagamentos_mais_usados = pagamentos[
    pagamentos == maior_frequencia
]

print("\n--- PAGAMENTOS MAIS UTILIZADOS POR REGIÃO ---")
print(pagamentos_mais_usados)
```

```python
# QUESTÃO 3

vendas_produtos = df.groupby("Produto")["Quantidade"].sum()

print("\n--- PRODUTO MAIS VENDIDO ---")
print(vendas_produtos.idxmax(), vendas_produtos.max())

print("\n--- PRODUTO MENOS VENDIDO ---")
print(vendas_produtos.idxmin(), vendas_produtos.min())
```

```python
# QUESTÃO

media_precos = df.groupby("Mercado")["Preco_Unitario"].mean()

print("\n--- PREÇO MÉDIO POR MERCADO ---")
print(media_precos.sort_values(ascending=False))
```

```python
# QUESTÃO 5s

df["Faturamento"] = df["Preco_Unitario"] * df["Quantidade"]

faturamento_total = df["Faturamento"].sum()

faturamento_mercado = df.groupby("Mercado")["Faturamento"].sum()

print("--- FATURAMENTO TOTAL DA REDE ---")
print(f"R$ {faturamento_total:.2f}")

print("\n--- FATURAMENTO POR MERCADO ---")
print(faturamento_mercado.sort_values(ascending=False))

print("\n--- MERCADO COM MAIOR FATURAMENTO ---")
print(faturamento_mercado.idxmax())
print(f"R$ {faturamento_mercado.max():.2f}")
```

```python
# QUESTÃO 6

df["Preco_Promocional"] = np.where(
    df["Quantidade"] >= 10,
    df["Preco_Unitario"] * 0.90,
    df["Preco_Unitario"]
)

df["Faturamento_Promocional"] = (
    df["Preco_Promocional"] * df["Quantidade"]
)

novo_faturamento = df["Faturamento_Promocional"].sum()

print("\n--- FATURAMENTO COM PROMOÇÃO ---")
print(f"R$ {novo_faturamento:.2f}")
```
